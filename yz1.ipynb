```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
text = "I love NLP. NLP is a branch of Artificial Intelligence. It deals with the study of making machines understand human language"

sentences = sent_tokenize(text)
words = word_tokenize(text)
```

```python
print("Sentence Tokenized: ")
for i, sentence in enumerate(sentences, 1):
    print("Sentence",i,":",sentence)

print("\nWords Tokenized: ") 
print(words)
```

```python
stemmer = PorterStemmer()
stemmed_words = []

for word in words:
    if word.isalpha():
        stemmed_words.append(stemmer.stem(word))
        print(word,"->",stemmed_words[-1])
```
