```python
import nltk

for package in ['punkt', 'punkt_tab', 'stopwords',
                    'omw-1.4', 'wordnet', 
                    'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng', 
                    'vader_lexicon']:
                    nltk.download(package, quiet=True)

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.tag import pos_tag
```

```python
text = """The students are studying natural language processing.
They are working on interesting projects and developing useful applications."""
```

```python
words = word_tokenize(text)

print("Tokenized Words:", words)
```

```python
pos_tagged = pos_tag(words)

for a,b in pos_tagged:
    print(a,"->",b)
```

```python
lemmatizer = WordNetLemmatizer()
lemmas = []

for word, tag in pos_tagged:
    if word.isalpha():
        if tag.startswith('J'):
            wordnet_tag = 'a'
        elif tag.startswith('V'):
            wordnet_tag = 'v'
        elif tag.startswith('R'):
            wordnet_tag = 'r'
        else:
            wordnet_tag = 'n'

        lemmas.append(lemmatizer.lemmatize(word, wordnet_tag))
        print(word,"->", lemmas[-1])
```
