```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import  RegexpParser
```

```python
text = "The student quickly solved the difficult problem in the laboratory"
```

```python
words = word_tokenize(text)
pos_tagged = pos_tag(words)
```

```python
grammar = r"""
NP : {<DT>?<RB.*>*<JJ.*>*<NN.*>+}
ADJP : {<RB.*>*<JJ.*>+}
ADVP : {<RB.*>+}
PP : {<IN><NP>}
VP : {<RB.*>*<VB.*>+}
"""

chunk_parser = RegexpParser(grammar)
tree = chunk_parser.parse(pos_tagged)

tree.pretty_print()
```

```python
# Semantic Chunking

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sentences = [
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks to learn from data.",
    "Cricket is a popular sport played with a bat and ball.",
    "Football is played between two teams."
]

vectorizer = TfidfVectorizer(stop_words='english')
vectors = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(vectors)
```

```python
visited = []
threshold = 0.10

for i in range(0, len(sentences)):
    if i in visited:
        continue

    print("\nChunk: ")
    print(sentences[i])
    visited.append(i)

    for j in range(i+1, len(sentences)):
        if similarity[i][j] >= threshold:
            print(sentences[j])
            visited.append(j)
```
