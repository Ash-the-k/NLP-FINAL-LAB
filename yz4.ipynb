```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
    "Natural Language processing is useful.",
    "Machine Learning is useful for language processing."
]
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
vector = vectorizer.fit_transform(documents)
```

```python
words = vectorizer.get_feature_names_out()
print("Learned Features:",words)
```

```python
print("Learned Weights:")
matrix = vector.toarray()
matrix
```

```python
print("Most frequent work in each Doc:")
for i in range(0, len(matrix)):
    print('Document',i+1,":",words[matrix[i].argmax()])
```

```python
# Semantic Parsing

sentence = "The student solved the difficult problem"
```

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import RegexpParser
```

```python
words = word_tokenize(sentence)
pos_tagged = pos_tag(words)
```

```python
grammar = r"""
NP : {<DT>?<RB.*>*<JJ.*>*<NN.*>+}
ADJP : {<RB.*>*<JJ.*>+}
ADVP : {<RB.*>+}
PP : {<IN><NP>}
VP : {<RB.*>*<VB.*>+}
"""

chunk_parser = RegexpParser(grammar)
tree = chunk_parser.parse(pos_tagged)

tree.pretty_print()
```
