```python
from collections import Counter
```

```python
text = "I love NLP and I love ML as I love NLP"

words = text.lower().split()
```

```python
words
```

```python
unigram = [(w) for w in words]
bigram = list(zip(words, words[1:]))
trigram = list(zip(words, words[1:], words[2:]))
```

```python
print("Unigram:")
print(unigram)

print("\nBigram:")
print(bigram)

print("\nTrigram:")
print(trigram)
```

```python
uni = Counter(unigram)
uni
```

```python
bi = Counter(bigram)
bi
```

```python
tri = Counter(trigram)
tri
```

```python
print("Unigram Probabilities:")
for u in uni:
    p = uni[u]/len(words)
    print(u,"->",round(p,2))
```

```python
print("Bigram Probabilities:")
for b in bi:
    p = bi[b]/uni[b[0]]
    print(b,"->",round(p,2))
```

```python
print("Trigram Probabilities:")
for t in tri:
    p = tri[t]/bi[t[:2]]
    print(t,"->",round(p,2))
```

```python
sequence = "I love NLP"
```

```python
sequence_words = sequence.lower().split()
```

```python
# Unigram

print("Unigram Probability")
print("-"*30)

uni_prob = 1

for w in sequence_words:
    uni_prob *= uni[w]/len(words)

print(sequence,"->", round(uni_prob,4))
```

```python
print("Bigram Probability")
print("-"*30)

bi_prob = 1

for i in range(1, len(sequence_words)):
    curr = sequence_words[i]
    prev = sequence_words[i-1]

    if uni[prev] == 0:
        bi_prob = 0
        break

    count = bi.get((prev, curr), 0)
    bi_prob *= count/uni[prev]

print(sequence,"->", round(bi_prob,4))
```

```python
print("Trigram Probability")
print("-"*30)

tri_prob = 1

for i in range(2, len(sequence_words)):
    prev_words = tuple(sequence_words[i-2: i])
    curr = sequence_words[i]

    if bi[prev_words] == 0:
        tri_prob = 0
        break

    count = tri.get(prev_words + (curr,),0)
    tri_prob *= count/bi[prev_words]

print(sequence,"->", round(tri_prob,4))
```
