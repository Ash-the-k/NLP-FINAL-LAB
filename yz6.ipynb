```python
import stanza
```

```python
stanza.download('en', processors='tokenize,ner')
```

```python
nlp_Pipeline = stanza.Pipeline('en','tokenize,ner')
```

```python
text = "Sundar Pichai is the CEO of Google. He visited Bangalore."
```

```python
doc = nlp_Pipeline(text)
```

```python
NER = doc.entities
```

```python
for i in NER:
    print(i.text,"->",i.type)
```
