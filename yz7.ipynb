```python

import re
import math
from collections import Counter
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')

# 1. Dataset
documents = [
    "Machine learning is a branch of artificial intelligence.",
    "Artificial intelligence uses machine learning algorithms.",
    "Machine learning algorithms are useful for data analysis.",
    "Natural language processing is an application of artificial intelligence."
]

# 2. Preprocessing
stop_words = set(stopwords.words('english'))
processed = []

for doc in documents:
    doc = doc.lower()
    words = re.findall(r'\b[a-z]+\b', doc)
    words = [w for w in words if w not in stop_words]
    processed.append(words)

# 3. Word Frequency
all_words = []

for doc in processed:
    all_words.extend(doc)

frequency = Counter(all_words)

print("WORD FREQUENCY:")
print(frequency)

print("\nMOST FREQUENT WORDS:")
highest = max(frequency.values())

for word, count in frequency.items():
    if count == highest:
        print(word, ":", count)

# 4. Calculate TF
tf_values = []

for doc in processed:
    count = Counter(doc)
    tf = {w: c / len(doc) for w, c in count.items()}
    tf_values.append(tf)

# 5. Calculate IDF
N = len(processed)
df = Counter()

for doc in processed:
    for word in set(doc):
        df[word] += 1

idf = {
    word: math.log(N / count)
    for word, count in df.items()
}

# 6. Calculate TF-IDF
tfidf_values = []
overall = Counter()

for tf in tf_values:
    tfidf = {}

    for word, score in tf.items():
        tfidf[word] = score * idf[word]
        overall[word] += tfidf[word]

    tfidf_values.append(tfidf)

# 7. Display document-wise TF-IDF
for i, scores in enumerate(tfidf_values):
    print(f"\nDOCUMENT {i+1}:")

    for word, score in sorted(
        scores.items(), key=lambda x: x[1], reverse=True
    ):
        print(word, ":", round(score, 4))

# 8. Overall TF-IDF Ranking
print("\nOVERALL TF-IDF RANKING:")

for word, score in overall.most_common():
    print(word, ":", round(score, 4))
```
